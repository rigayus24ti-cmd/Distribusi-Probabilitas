```python
# Menghitung probabilitas sederhana

jumlah_seluruh = 6
jumlah_genap = 3

probabilitas = jumlah_genap / jumlah_seluruh

print("Jumlah seluruh kemungkinan:", jumlah_seluruh)
print("Jumlah kejadian genap:", jumlah_genap)
print("Probabilitas:", probabilitas)
print("Persentase:", probabilitas * 100, "%")
```

```python
# Menghitung probabilitas komplemen

prob_gangguan = 0.15

prob_tidak_gangguan = 1 - prob_gangguan

print("Probabilitas gangguan:", prob_gangguan)
print("Probabilitas tidak gangguan:", prob_tidak_gangguan)
print("Persentase server normal:", prob_tidak_gangguan * 100, "%")
```

```python
# Probabilitas gabungan kejadian independen

server_1 = 0.1
server_2 = 0.2

prob_bersamaan = server_1 * server_2

print("Gangguan server 1:", server_1)
print("Gangguan server 2:", server_2)
print("Probabilitas keduanya terganggu:", prob_bersamaan)
print("Persentase:", prob_bersamaan * 100, "%")
```

```python
# Probabilitas bersyarat

pengguna_2fa = 40
berhasil_verifikasi = 30

probabilitas = berhasil_verifikasi / pengguna_2fa

print("Jumlah pengguna 2FA:", pengguna_2fa)
print("Berhasil verifikasi:", berhasil_verifikasi)
print("Probabilitas bersyarat:", probabilitas)
print("Persentase:", probabilitas * 100, "%")
```

```python
# Menghitung Teorema Bayes

prob_spam = 0.20
prob_normal = 1 - prob_spam

deteksi_spam = 0.90
false_positive = 0.05

pembilang = deteksi_spam * prob_spam

penyebut = (
    deteksi_spam * prob_spam
    + false_positive * prob_normal
)

hasil = pembilang / penyebut

print("Probabilitas email benar-benar spam:", round(hasil, 4))
print("Persentase:", round(hasil * 100, 2), "%")
```

```python
import math

jumlah_anggota = 5
jumlah_dipilih = 2

kombinasi = math.comb(
    jumlah_anggota,
    jumlah_dipilih
)

permutasi = math.perm(
    jumlah_anggota,
    jumlah_dipilih
)

print("Jumlah kombinasi:", kombinasi)
print("Jumlah permutasi:", permutasi)
```

```python
import numpy as np

# Membuat generator angka acak
rng = np.random.default_rng(42)

# Simulasi 10.000 lemparan dadu
hasil = rng.integers(1, 7, size=10000)

print("Jumlah percobaan:", len(hasil))

for angka in range(1, 7):
    frekuensi = np.sum(hasil == angka)
    probabilitas = frekuensi / len(hasil)

    print(
        f"Angka {angka}: "
        f"Frekuensi = {frekuensi}, "
        f"Probabilitas = {probabilitas:.4f}"
    )
```

```python
# Aturan penjumlahan probabilitas

android = 0.60
ios = 0.30
keduanya = 0.10

probabilitas = android + ios - keduanya

print("Probabilitas Android atau iOS:", probabilitas)
print("Persentase:", probabilitas * 100, "%")
```

```python
from scipy.stats import bernoulli

p = 0.8

mean, variance = bernoulli.stats(
    p,
    moments="mv"
)

print("Nilai harapan:", mean)
print("Varians:", variance)

print("Peluang berhasil:", bernoulli.pmf(1, p))
print("Peluang gagal:", bernoulli.pmf(0, p))
```

```python
from scipy.stats import binom

n = 10
p = 0.8
x = 8

prob = binom.pmf(x, n, p)

mean = binom.mean(n, p)
variance = binom.var(n, p)

print("Probabilitas tepat 8 berhasil:", prob)
print("Nilai harapan:", mean)
print("Varians:", variance)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.8

x = np.arange(0, n + 1)
y = binom.pmf(x, n, p)  # Sediakan parameter n dan p di sini

# Contoh jika ingin menampilkan grafiknya:
plt.bar(x, y)
plt.xlabel('x')
plt.ylabel('Probability')
plt.title('Distribusi Binomial')
plt.show()
```
